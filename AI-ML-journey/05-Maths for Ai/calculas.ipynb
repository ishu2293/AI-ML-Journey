```python
# Plot y = √x
import numpy as np
import matplotlib.pyplot as plt

# Define x values (x >= 0 for sqrt)
x = np.linspace(0, 100, 100_000)
y = np.sqrt(x)

# Plot
plt.plot(x, y, label="y = √x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = √x")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Create x values from -2π to 2π
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)
y = np.sin(x)

# Plot the graph
plt.plot(x, y, label='y = sin(x)', color='blue')

# Highlight x=0 and y=0 lines
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of y = sin(x)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Create x values from -2π to 2π
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)

# Scalar multiplication
y = 2*(np.sin(x))
y2 = -2*(np.sin(x))

# Plot the graph
plt.plot(x, y, label='y = 2*sin(x)', color='blue')
plt.plot(x, y2, label='y = -2*sin(x)', color='red')

# Highlight x=0 and y=0 lines
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of y = 2*sin(x) vs y = -2*sin(x)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Create x values from -2π to 2π
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)

# Scalar Addition
y = 2+(np.sin(x))
y2 = 2-(np.sin(x))

# Plot the graph
plt.plot(x, y, label='y = 2+sin(x)', color='blue')
plt.plot(x, y2, label='y = 2-sin(x)', color='red')

# Highlight x=0 and y=0 lines
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of y = 2+sin(x) vs y = 2-sin(x)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Create x values from -2π to 2π
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)

y = np.sin(x)
# Scalar multiplication in input
y2 = np.sin(2*x)

# Plot the graph
plt.plot(x, y, label='y = sin(x)', color='blue', alpha = 0.2)
plt.plot(x, y2, label='y = sin(2x)', color='red')

# Highlight x=0 and y=0 lines
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of y = sin(x) vs y = sin(2x)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Create x values from -2π to 2π
x = np.linspace(-2 * np.pi, 2 * np.pi, 100)

y = x**2
# Scalar addition in input
y2 = (x + 2)**2

# Plot the graph
plt.plot(x, y, label='y = x**2', color='blue', alpha = 0.2)
plt.plot(x, y2, label='y = (x + 2)**2', color='red')

# Highlight x=0 and y=0 lines
plt.axvline(0, color='black', linewidth=1.5, linestyle='--', label='x = 0')
plt.axhline(0, color='black', linewidth=1.5, linestyle='--', label='y = 0')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of y = x**2 vs y = (x + 2)**2')
plt.grid(True)
plt.legend()

plt.show()
```
