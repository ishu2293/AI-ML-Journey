```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# ques - 1
scores = np.random.normal(loc=70, scale=10, size=200)
plt.hist(scores, bins = 10, 
         edgecolor = "black",
        linewidth = 2)
plt.axvline(x = scores.mean(), 
            color = "yellow", 
            linestyle = "--",
            linewidth = 2,
            label = "Mean"
           )
```

```python
# ques - 2
penguins = sns.load_dataset("penguins")
penguins.head()
sns.relplot(
    data = penguins,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species"
)
```

```python
# ques - 3
tips = sns.load_dataset("tips")
tips
sns.boxplot(
    data = tips,
    x = "total_bill",
    y = "day",
    hue = "sex",
    
)
```

```python
# ques - 4
fig, ax = plt.subplots()
tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker",
    ax = ax
)

ax.set_title("Total bill v/s the tip given")
ax.set_xlabel("Total_bill")
ax.set_ylabel("Tip")
```

```python
tips = sns.load_dataset("tips")
tips
tips_p = tips.pivot_table(index = "day", columns = "time", values = "tip", aggfunc = "mean", observed=False)
tips_p
sns.heatmap(
    data = tips_p,
    annot = True 
)
```
