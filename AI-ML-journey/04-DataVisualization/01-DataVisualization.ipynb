```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.plot(years, revenue)
plt.title("Oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in dollars")
```

```python
non_oscar_movies = {"Slundog Millionare", "Avatar", "Inception", "Hugo", "Lincoln"}
n_revenue = [378, 2788, 829, 185, 275]

plt.plot(years, revenue, label = "oscar_movies")
plt.plot(years, n_revenue, label = "non-oscar_movies")
plt.title("Oscar movies v/s non_oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in dollars")
plt.legend()
```

```python
# Bar plot - Bar chart
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.bar(years, revenue, color = "purple")
plt.title("revenue in each year for oscar movies")
plt.xlabel("years")
plt.ylabel("Revenue in dollars")

# x pos, y pos, value, alignment
for i in range(len(years)):
    plt.text(years[i], revenue[i]+10, str(revenue[i]), ha = "center")

plt.ylim(0, max(revenue)+70)
```

```python
import numpy as np

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]
non_oscar_movies = {"Slundog Millionare", "Avatar",
                    "Inception", "Hugo", "Lincoln"}
n_revenue = [378, 2788, 829, 185, 275]

x = np.arange(len(years)) 
width = 0.4

plt.bar(x-width/2, revenue, width, label = "Oscar movies")
plt.bar(x+width/2, n_revenue, width, label = "Non-Oscar movies")
plt.xticks(x, years)
plt.title("Oscar v/s Non-Oscar movies")
plt.xlabel("years")
plt.ylabel("Revenue in dollars")
plt.legend()
```

```python
# Scatter plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# plt.scatter(age, blood_pressure, color="red")
plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()

# Customizing plots
colors = ["green" if bp<135 else "red" for bp in blood_pressure]
sizes = blood_pressure
plt.scatter(age, blood_pressure, color=colors, s=sizes, marker='^', alpha=0.7)

# plt.scatter(age, blood_pressure, c=blood_pressure, cmap="plasma")
plt.scatter(age, blood_pressure, c=blood_pressure, cmap="OrRd")# makes more sense
# plt.scatter(age, blood_pressure, c=[5]*len(age), cmap="plasma")
plt.colorbar(label="BP")

for i in range(len(age)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i])) #(text, xy pos, text pos)
plt.xlim(min(age), max(age)+10)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

# Plots
plt.scatter(winter_temp, winter_humidity, color="blue", label="Winter")
plt.scatter(summer_temp, summer_humidity, color="orange", label="Summer")
plt.title("Summer v/s Winter Data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses)

# set style
plt.style.use("ggplot")

# add %-age of each slice
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%')

#set colors
colors = ["purple", "green", "blue", "pink", "yellow"]
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', colors=colors)

# wedgeprops
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', 
        wedgeprops={"edgecolor":"black", 
                    "linewidth": 2, 
                    "linestyle":"-.", 
                    "fill":False}
       )
```
