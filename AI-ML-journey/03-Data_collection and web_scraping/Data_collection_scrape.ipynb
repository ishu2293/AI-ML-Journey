```python
from bs4 import BeautifulSoup
with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
# soup
```

```python
soup.find("h3")
```

```python
soup.h3
```

```python
all_h3 = soup.find_all("h3")
type(all_h3)
```

```python
all_countries = []
for h3 in all_h3:
    name = h3.get_text(strip = True) #country names
    # print(h3.find_parent("div")["class"])
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip = True) #country population

    all_countries.append([name, population])
```

```python
all_countries
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns = ["Name", "Population"])
df
```

```python
df.to_csv("scraped_data/data1_cleaned.csv", index = True)
```
