```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
page_c = 1
while (page_c < 6):
    url = f"https://books.toscrape.com/catalogue/page-{page_c}.html"
    res = requests.get(url)
    
    with open(f"hw-web_scrape_books/book{page_c}.html", "w", encoding = "utf-8") as f:
        f.write(res.text)
        print(f"Downloaded book page {page_c}")
    page_c = page_c + 1
        
```
