```python
import pandas as pd
```

```python
#DataFrame

info = {
    "Name": ["Ishwari", "Pranita", "Saee", "Vaishnavi"],
    "Roll_No": [44, 10, 48, 58]
}
df = pd.DataFrame(info)
df
```

```python
#Series with index
s = pd.Series([23, 24, 18, 19])
print(s)
print(s[0])
print(s[3])
```

```python
#Series with label

s = pd.Series([23, 24, 18, 19], index = ["Ishwari", "Vaishnavi", "Saee", "Pranita"])
print(s)
print(s["Ishwari"])
print(s[1])
```

```python
#DataFrame

#using dictionary
info = {
    "Name": ["A", "B", "C"],
    "Age": [19, 20, 21]
}
df = pd.DataFrame(info)
df
```

```python
#using list of lists
infol = [["A", 23], ["B", 20], ["C", 21]]
df = pd.DataFrame(infol, columns = ["Name", "Age"])
df
```

```python
#using numpy array
import numpy as np
arr = np.array([["A", 26], ["B", 30], ["C", 21]])
df = pd.DataFrame(arr, columns = ["Name", "Age"])
df
```
